```python
meat_grouped = meat_consumption.groupby(["Entity", "Code", "Continent", "Year", "Category"], as_index=False)["Value"].sum()
meat_grouped.shape   # expected (14427, 6)
```

```python
meat_grouped[(meat_grouped["Code"] == "ESP") & (meat_grouped["Year"] == 2023)]   
```

```python
meat_cols = ["Poultry", "Beef and buffalo", "Sheep and goat", "Pork", "Other meats", "Fish and seafood"]

wide = meat_grouped.pivot(index=["Entity", "Code", "Continent", "Year"], columns="Category", values="Value").reset_index()
wide.columns.name = None
wide = wide[["Entity", "Code", "Continent", "Year"] + meat_cols]
wide.shape   
```

```python
wide[meat_cols].isna().sum()  
```

```python
wide[meat_cols] = wide[meat_cols].fillna(0).round(2)
wide[meat_cols].isna().sum()   # all 0 now
```

```python
wide["Complete_series"] = wide.groupby("Code")["Year"].transform("nunique") == wide["Year"].nunique()
wide.groupby("Code")["Complete_series"].first().value_counts()   # True 168, False 10
```

## Dataset 1: continents 

```python
continent_codes = {"Africa": "OWID_AFR", "Asia": "OWID_ASI", "Europe": "OWID_EUR",
                   "North America": "OWID_NAM", "Oceania": "OWID_OCE", "South America": "OWID_SAM"}

complete = wide[wide["Complete_series"]]
meat_continents = complete.groupby(["Continent", "Year"])[meat_cols].mean().round(2).reset_index()
meat_continents["Code"] = meat_continents["Continent"].map(continent_codes)
meat_continents = meat_continents.rename(columns={"Continent": "Entity"})
meat_continents = meat_continents[["Entity", "Code", "Year"] + meat_cols]
meat_continents.shape  
```

```python
meat_continents[meat_continents["Year"] == 2023]
```

## Dataset 2: the 9 countries

```python
wanted = ["IRN", "NLD", "UKR", "FRA", "ESP", "ITA", "ARG", "GBR", "PRT", "BRA"]

meat_countries = wide[wide["Code"].isin(wanted)].drop(columns=["Continent", "Complete_series"])
meat_countries = meat_countries.sort_values(["Entity", "Year"]).reset_index(drop=True)
print(meat_countries.shape)                 
print(set(wanted) - set(meat_countries["Code"]))  
```

```python
meat_columns = ["Poultry", "Beef and buffalo", "Sheep and goat", "Pork", "Other meats", "Fish and seafood"]

meat_countries["Total"] = meat_countries[meat_columns].sum(axis=1)

total = (
    meat_countries[
        (meat_countries["Code"].isin(wanted)) &
        (meat_countries["Year"] == 2023)
    ][["Code"] + meat_columns + ["Total"]]
    .copy()
)

total = total.sort_values("Total", ascending=False)

meat_countries[meat_countries["Year"] == 2023].sort_values("Total", ascending=False)
```
