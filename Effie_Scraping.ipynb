```python
cells = rows[1].find_all(["th", "td"])
print(len(cells))

row_values = [cell.get_text(strip=True) for cell in cells]
print(row_values)
```

```python
header_cells = rows[0].find_all(["th", "td"])
columns = [cell.get_text(strip=True) for cell in header_cells]
```

```python
columns
```

# Building the Dataframe table

```python
data = []

for row in rows[1:]: #1: to skip the header (make sure you put the : after and not before the num)
    cells = row.find_all(["th", "td"])
    row_values = [cell.get_text(strip=True) for cell in cells]
    data.append(row_values)
```

```python
len(data)
```

```python
data
```

```python
df = pd.DataFrame(data, columns=columns)
df.head()
```

```python
df
```
